```python
import json
import time

import joblib
import pandas as pd
from pathlib import Path
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.pipeline import Pipeline

import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder, OrdinalEncoder
```

```python

#####################
#CONFIG
######################

NOTEBOOK_DIR = Path.cwd()

# If notebook is inside PROJECT_ROOT/notebooks:
PROJECT_ROOT = NOTEBOOK_DIR.parent.parent

DATA_DIR = PROJECT_ROOT / "project-7-files"
OUTPUT_DIR = PROJECT_ROOT / "outputs"
FIGURES_DIR = OUTPUT_DIR / "figures"
MODELS_DIR = OUTPUT_DIR / "models"

for d in (OUTPUT_DIR, FIGURES_DIR, MODELS_DIR):
    d.mkdir(parents=True, exist_ok=True)

RANDOM_STATE = 42
ID_COL = "Pkey"
TARGET_COL = "target"
N_CV_FOLDS = 5

# Approximate Paris coordinates in the Lambert-93 projection (in metres).
# Used to derive a "distance to Paris" feature as a cheap proxy for
# centrality / access to the main economic hub.
PARIS_LAMBERT93_X = 652_709.0
PARIS_LAMBERT93_Y = 6_862_008.0
```

```python
####################
#Modeling
####################

ID_COL = "Pkey"
TARGET_COL = "target"

DROP_COLS = [
    "INSEE_CODE",
    "Nom de la commune",
    "Nom du département",
    "Nom de la région",
    "club_federation",
    "X",
    "Y",
]

CATEGORICAL_COLS = [
    "highest_credential",
    "Job_42",
    "HOUSEHOLD",
    "Sex",
    "Act",
    "EMP_CONTRACT",
    "Eco_sect",
    "Employee_count",
    "Job_dep",
    "COMPANY_CATEGORY",
    "type_of_contract",
    "labor_force_status",
    "Job_condition",
    "Job_description_N2",
    "Job_description_N1",
    "Former_job_42",
    "FORMER_EMP_CONTRACT",
    "Eco_sect_former_job",
    "former_dep",
    "Job_dep_former_job",
    "type_of_contract_former_job",
    "labor_force_status_former_job",
    "Job_condition_former_job",
    "COMPANY_CATEGORY_former_job",
    "Employee_count_former_job",
    "Former_job_description_N2",
    "Former_job_description_N1",
    "club_category",
    "city_type",
    "dep",
    "Reg",
]

BOOLEAN_COLS = [
    "IS_STUDENT",
    "has_emp_contract",
    "has_job_detail",
    "has_retired_former",
    "has_retired_job_detail",
    "has_pension",
    "has_club",
]

NUMERIC_COLS = [
    "Age_2020",
    "PAY",
    "working_hours",
    "retirement_age",
    "working_hours_former_job",
    "retirement_pay",
    "LAT",
    "long",
    "INHABITANTS",
    "distance_to_paris_km",
    "log_inhabitants",
]

#inspect all columns in the DF and determine which feature belongs on each group
def get_feature_columns(df):
    """Restrict the column groups above to what is actually in `df`."""
    cols = set(df.columns) - {ID_COL, TARGET_COL} - set(DROP_COLS)
    categorical = [c for c in CATEGORICAL_COLS if c in cols]
    boolean = [c for c in BOOLEAN_COLS if c in cols]
    numeric = [c for c in NUMERIC_COLS if c in cols]
    known = set(categorical + boolean + numeric)
    leftover = cols - known
    if leftover:
        raise ValueError(f"Unclassified columns, update common/modeling.py: {leftover}")
    return categorical, boolean, numeric


#converts categorical columns into strings while preserving missing values.
def _stringify(X):
    X = X.copy()
    for col in X.columns:
        na_mask = X[col].isna()
        X[col] = X[col].astype(str)
        X.loc[na_mask, col] = np.nan
    return X


```

```python
####################
#Random forest
####################

#-----------------
#Define the hyperparameter grid
#----------------

PARAM_GRID = {
    "model__n_estimators": [300], #Number of decision trees in the forest
    "model__max_depth": [None, 20, 30], #Maximum depth of each tree; None allows growth until other stopping conditions apply
    "model__min_samples_leaf": [1, 5, 20], #Minimum number of training samples required in a leaf
    "model__max_features": ["sqrt", 0.3], #Number or proportion of features considered when searching for the best split
}

#---------------------
#Random forest processor
#---------------------
def random_forest_preprocessor(df):
    categorical, boolean, numeric = get_feature_columns(df)

    categorical_pipe = Pipeline(
        steps=[
            ("stringify", FunctionTransformer(_stringify, feature_names_out="one-to-one")),
            ("impute", SimpleImputer(strategy="constant", fill_value="__missing__")),
            ("encode", OneHotEncoder(handle_unknown="ignore")),
        ]
    )
    numeric_pipe = Pipeline(steps=[("impute", SimpleImputer(strategy="median"))])
    #missing values are replaced with the median of the corresponding feature.

    #Apply categrorical pipeline to categorical columns and numerical pipeline to numerical columns (and boolean too)
    preprocessor = ColumnTransformer(
        transformers=[
            ("cat", categorical_pipe, categorical),
            ("num", numeric_pipe, numeric + boolean),
        ]
    )
    return preprocessor, categorical, boolean, numeric
```

```python
#----------------------
#Create pipeline to use
#----------------------

def main():
    #Load dataset
    learn = pd.read_parquet(OUTPUT_DIR / "learn_features.parquet")
    y = learn[TARGET_COL] #contains the correct outcomes the model must learn to predict.
    X = learn.drop(columns=[TARGET_COL]) #contains the input features used to make predictions.

#v1
    preprocessor, categorical, boolean, numeric = random_forest_preprocessor(X) #load function created previously
#v2
#   preprocessor, categorical, boolean, numeric = random_forest_preprocessor_v2(X) #load function created previously
    print(f"{len(categorical)} categorical, {len(boolean)} boolean, {len(numeric)} numeric features")

    #Build the actual pipeline
    pipe = Pipeline(
        steps=[
            ("prep", preprocessor),
            ("model", RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1)),
        ]
    )


    #Cross validation
    cv = StratifiedKFold(n_splits=N_CV_FOLDS, shuffle=True, random_state=RANDOM_STATE)

# Five-fold stratified cross-validation splits the dataset into five parts.
# Each part serves as the validation set once, while the other four parts are used for training.
# The stratification attempts to preserve the proportion of each target class in each fold. 

    #Running gridsearchCV
    search = GridSearchCV(
        pipe,
        PARAM_GRID,
        scoring="roc_auc",
        cv=cv,
        n_jobs=-1,
        refit=True,
        verbose=2,
    )

#GridSearchCV tests each of the 18 parameter configurations across all five folds. (18x5=90)


    start = time.time()
    search.fit(X, y)
    elapsed = time.time() - start
    print(f"Grid search done in {elapsed:.1f}s")
    print("Best params:", search.best_params_)
    print("Best CV ROC AUC:", search.best_score_)

    cv_results = pd.DataFrame(search.cv_results_)
    cv_results.to_csv(OUTPUT_DIR / "rf_cv_results.csv", index=False)

    joblib.dump(search.best_estimator_, MODELS_DIR / "random_forest.joblib")

    with open(OUTPUT_DIR / "rf_best_params.json", "w") as f:
        json.dump(
            {"best_params": search.best_params_, "best_cv_roc_auc": search.best_score_},
            f,
            indent=2,
        )

```

```python
#Execute the pipeline

if __name__ == "__main__":
    main()
```

```python
# =====================================================================================
# [ADDED BY CLAUDE] Corrected version of random_forest_preprocessor
# =====================================================================================
# Problem with the original version (cell "#Modeling" above):
#   numeric_pipe = SimpleImputer(strategy="median") fills EVERY numeric NaN with the median.
#   But most NaNs in PAY, working_hours, retirement_age, retirement_pay, ... are not "unknown":
#   the value does not exist (e.g. a student has no salary, an employed person has no pension).
#   With the median, a student gets the median salary of workers, i.e. a fake but plausible value.
#
# Fix:
#   - "Not applicable" numeric columns (they come from the job / retired tables) are filled
#     with a sentinel value (-1). It is impossible for a real value (all of them are > 0),
#     so the trees can isolate it with a single split.
#   - add_indicator=True also adds a 0/1 column "is this value missing?" for each of them.
#   - The other numeric columns (age, geography) have no NaN; they keep the median
#     imputation as a safety net for unexpected NaNs in new data.
#   - Categorical and boolean handling is unchanged.
#
# The original random_forest_preprocessor is NOT modified. This is a new function
# (random_forest_preprocessor_v2) so the results already obtained stay valid.

from sklearn.model_selection import cross_val_score

NOT_APPLICABLE_NUMERIC_COLS = [
    "PAY",                       # only people in the job table (employed)
    "working_hours",             # only people in the job table
    "retirement_age",            # only retired people
    "working_hours_former_job",  # only retired people
    "retirement_pay",            # only retired people
]


def random_forest_preprocessor_v2(df):
    categorical, boolean, numeric = get_feature_columns(df)
    not_applicable = [c for c in numeric if c in NOT_APPLICABLE_NUMERIC_COLS]
    other_numeric = [c for c in numeric if c not in NOT_APPLICABLE_NUMERIC_COLS]

    categorical_pipe = Pipeline(
        steps=[
            ("stringify", FunctionTransformer(_stringify, feature_names_out="one-to-one")),
            ("impute", SimpleImputer(strategy="constant", fill_value="__missing__")),
            ("encode", OneHotEncoder(handle_unknown="ignore")),
        ]
    )
    # Sentinel -1 + missing indicator for values that do not exist
    not_applicable_pipe = Pipeline(
        steps=[("impute", SimpleImputer(strategy="constant", fill_value=-1, add_indicator=True))]
    )
    # Median only where a NaN would really mean "unknown"
    numeric_pipe = Pipeline(steps=[("impute", SimpleImputer(strategy="median"))])

    preprocessor = ColumnTransformer(
        transformers=[
            ("cat", categorical_pipe, categorical),
            ("num_not_applicable", not_applicable_pipe, not_applicable),
            ("num", numeric_pipe, other_numeric + boolean),
        ]
    )
    return preprocessor, categorical, boolean, numeric


# -------------------------------------------------------------------------------------
# Quick comparison: original vs corrected preprocessing, with the best hyperparameters
# found by the grid search (rf_best_params.json) and the same 5 CV folds.
# -------------------------------------------------------------------------------------
learn = pd.read_parquet(OUTPUT_DIR / "learn_features.parquet")
y = learn[TARGET_COL]
X = learn.drop(columns=[TARGET_COL])

with open(OUTPUT_DIR / "rf_best_params.json") as f:
    best = json.load(f)["best_params"]

cv = StratifiedKFold(n_splits=N_CV_FOLDS, shuffle=True, random_state=RANDOM_STATE)

for name, make_prep in [
    ("original (median)", random_forest_preprocessor),
    ("corrected (sentinel -1 + indicator)", random_forest_preprocessor_v2),
]:
    prep, *_ = make_prep(X)
    pipe = Pipeline(
        steps=[
            ("prep", prep),
            ("model", RandomForestClassifier(
                random_state=RANDOM_STATE,
                n_jobs=-1,
                n_estimators=best["model__n_estimators"],
                max_depth=best["model__max_depth"],
                min_samples_leaf=best["model__min_samples_leaf"],
                max_features=best["model__max_features"],
            )),
        ]
    )
    scores = cross_val_score(pipe, X, y, scoring="roc_auc", cv=cv)
    print(f"{name:38s} CV ROC AUC = {scores.mean():.4f} ± {scores.std():.4f}  (folds: {np.round(scores, 4)})")

# [ADDED BY CLAUDE] Result obtained (see output below):
#   original (median)                    CV ROC AUC = 0.7937 ± 0.0035
#   corrected (sentinel -1 + indicator)  CV ROC AUC = 0.7938 ± 0.0036
# -> Same performance. The has_* flags already told the trees who has no job / no pension,
#    so the fake median values did not hurt the score. The corrected version is still
#    methodologically cleaner (no invented values) and easier to justify in the report.
```
